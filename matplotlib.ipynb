```python
import matplotlib.pyplot  as plt
```

```python
x=[1,2,3,4]
y=[2,4,6,8]
plt.plot(x,y,color='red',linestyle='--',marker='o')
plt.title('My plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x=[1,2,3,4]
y1=[3,4,6,7]
y2=[4,7,8,5]
plt.plot(x,y1,label='Squares')
plt.plot(x,y2,label='Linear')
plt.legend()
plt.show()
```

```python
a=['A','B','C']
b=[5,7,9]
plt.bar(a,b)
plt.show()
```

```python
data=[2,2,6,9,5,6,7,3,3,3]
plt.hist(data,bins=5)
plt.show()
```

```python
x=[3,4,5,6]
y=[4,3,6,7]
plt.scatter(x,y)
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [1, 2, 3])
plt.subplot(1,3,2)
plt.plot([1,2,3],[1,3,3])
plt.show()
```
